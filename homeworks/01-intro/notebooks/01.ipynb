```
new_df = df[df["origin"] == "Asia"]
new_df.shape
```

```
new_df_sub = new_df[["vehicle_weight", "model_year"]]
print(new_df_sub)
```

```
specific_sub = new_df_sub.head(7)
print(specific_sub)
```

```
X = specific_sub.to_numpy()
print(X)
```

```
XTX_inv = np.linalg.inv(X.T @ X)
print(XTX_inv)
```

```
y = [ 1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_inv @ X.T @ y
print(w.sum())
```
